```python
from sklearn.model_selection import train_test_split
Y=df['class']
X=df.drop('class', axis=1)
X_train, X_test, y_train, y_test=train_test_split(X, Y, random_state=42, test_size=0.2)
print(f"Test Size: {len(X_test)}")
```

```python
from sklearn.linear_model import LogisticRegression
log=LogisticRegression()
log
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score

log.fit(X_train, y_train)
y_pred=log.predict(X_test)
print("Baseline Logistic Regression\nAccuracy: ", accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_validate
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#----------Hyper parameter tuning---------------------

param_grid={
    'penalty':['l1', 'l2'],
    'C':[0.01, 0.1, 1, 10, 100],
    'solver':['saga','liblinear']
}

search=GridSearchCV(estimator=LogisticRegression(),
                   param_grid=param_grid,
                   cv=5,
                   scoring='accuracy',
                   n_jobs=-1
                   )

search.fit(X_train, y_train)

print(f"Best Parameters of Logistic Regression: ")
print(search.best_params_)
print("Best Accuracy Score: ")
print(search.best_score_)
```

```python
import time

#----------Scoring-----------------
best_model=search.best_estimator_
start=time.perf_counter()
best_model.fit(X_train, y_train)
end=time.perf_counter()
duration=end-start
print("Training time: ", duration)

y_pred=best_model.predict(X_test)

acc=accuracy_score(y_test, y_pred)
rec=recall_score(y_test, y_pred)
prec=precision_score(y_test, y_pred)
f1=f1_score(y_test, y_pred)

print("Accuracy: ", acc)
print("Recall: ", rec)
print("Precision: ", prec)
print("F1 score: ", f1)
```

```python
#------------Cross Validation-------------
cv_results = cross_validate(
        search.best_estimator_,
        X_train,
        y_train,
        cv=5,
        scoring='accuracy'
)
cv_acc   = cv_results['test_score']

for i in range(5):
    print(f"K={i}    Score={cv_acc[i]}")

print("\n\nMean score: ", cv_acc.mean())
```

```python
from sklearn import svm
SVM=svm.SVC()
SVM
```

```python
SVM.fit(X_train, y_train)
y_pred=SVM.predict(X_test)
print("Baseline Support Vector Machine Classifier\nAccuracy: ", accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_validate
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#----------Hyper parameter tuning---------------------

param_grid={
    'kernel':['linear', 'poly', 'rbf','sigmoid'],
    'C':[0.1, 1, 10, 100],
    'degree':[2,3,4]
}

search=GridSearchCV(estimator=svm.SVC(),
                   param_grid=param_grid,
                   cv=5,
                   scoring='accuracy',
                   n_jobs=-1
                   )

search.fit(X_train, y_train)

print(f"Best Parameters of SVM Classification: ")
print(search.best_params_)
print("Best Accuracy Score: ")
print(search.best_score_)
```

```python
def svm_kernel_metrics(metrics_df, kernel_i):
    model=SVM(kernel=kernel_i)
    start=time.perf_counter()
    model.fit(X_train, y_train)
    y_pred=model.predict(X_test)
    end=time.perf_counter()
    duration=end-start
    t=y_test
    p=y_pred
    length=len(metrics_df)
    metrics_df.loc[length+1]=([kernel_i, accuracy_score(t, p), f1_score(t,p), duration])
```

```python
cols=['kernel', 'accuracy', 'f1_score', 'time_taken']
svm_metrics = pd.DataFrame(columns=cols)
```

```python
kernels=['linear', 'poly', 'rbf','sigmoid']
for kernel in kernels:
    svm_kernel_metrics(kernel)
svm_metrics
```
